**Checking And Droping all the Null Value**

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

**Checking for the duplicadted values**

```python
df.duplicated().sum()
```

**Converting Column's datatype in proper format**

```python
#Converting release date from object to datetime
df['album_release_date']=pd.to_datetime(df['album_release_date'], format='mixed')
```

```python
#Converting genre string list into actual Python List
df['artist_genres'] = df['artist_genres'].apply(eval)
```

```python
#Converting album_type to category
df['album_type'] = df['album_type'].astype('category')
```

```python
df.info()
```

**Extract Year, Month, Day, and Day Name from album_release_date**

```python
df['release_year'] = df['album_release_date'].dt.year
df['release_month'] = df['album_release_date'].dt.month
df['release_day'] = df['album_release_date'].dt.day
df['release_day_name'] = df['album_release_date'].dt.day_name()
```

```python
df.head()
```

**Adding Duration of tracks in minutes and in MM:SS format**

```python
df['track_duration_min'] = (df['track_duration_ms'] / 60000).round(2)
```

```python
#Adding a column for duration format as 'MM:SS'
minutes = df['track_duration_ms'] // 60000
seconds = (df['track_duration_ms'] % 60000) // 1000
df['track_duration_mmss'] = minutes.astype(str).str.zfill(2) + ':' + seconds.astype(str).str.zfill(2)
```

```python
df.head(7)
```

**Creating the popularity tier column and the followers bin**

```python
pop_bins = [-1, 33, 66, 100]
pop_labels = ['Low', 'Medium', 'High']

df['track_popularity_tier'] = pd.cut(df['track_popularity'], bins=pop_bins, labels=pop_labels)
df['artist_popularity_tier'] = pd.cut(df['artist_popularity'], bins=pop_bins, labels=pop_labels)
```

```python
#Creating artist Followers bin column
follower_bins = [-1, 100000, 1000000, 10000000, float('inf')]
follower_labels = ['< 100K', '100K–1M', '1M–10M', '10M+']

df['artist_follower_bin'] = pd.cut(df['artist_followers'], bins=follower_bins, labels=follower_labels)
```
